```python
# PRACTICE QUESTION — Topic 4

# Step 1: Collect the student's scores
exam_score = float(input("Enter your exam score: "))
coursework_score = float(input("Enter your coursework score: "))
attendance = float(input("Enter your attendance percentage: "))

# Step 2: Calculate the final score
# Exam contributes 60% and coursework contributes 40%.
final_score = (exam_score * 0.60) + (coursework_score * 0.40)

# Step 3: Apply a 5% bonus if attendance is 90% or above
bonus_applied = False

if attendance >= 90:
    final_score += 5
    bonus_applied = True

# Step 4: Determine if the student passed
# The student must have a final score of at least 50 and attendance of at least 60%.
if final_score >= 50 and attendance >= 60:
    result = "Pass"
else:
    result = "Fail"

# Step 5: Display the results
print("\n" + "=" * 45)
print("         GRADE CALCULATOR")
print("=" * 45)
print("Final Score:", round(final_score, 2))
print("Pass/Fail Status:", result)
print("Attendance Bonus Applied:", bonus_applied)
print("=" * 45)
```

```python
# PRACTICE QUESTION — Topic 6
# Write your Number Pattern Printer here

# Step 1: Ask the user for a number between 1 and 20
number = int(input("Enter a number between 1 and 20: "))

# Validate the input using a while loop
while number < 1 or number > 20:
    print("Invalid input. Please enter a number between 1 and 20.")
    number = int(input("Enter a number between 1 and 20: "))

# Step 2: Print the descending number triangle
print("\nNumber Pattern:")

for row in range(number, 0, -1):
    for value in range(number, number - row, -1):
        print(value, end=" ")
    print()

# Step 3: Print only numbers that are NOT multiples of N
print("\nNumbers that are NOT multiples of", number)

count = 0
for i in range(1, (number * 2) + 1):

    if i % number == 0:
        continue

    print(i)

    count += 1

    if count == 5:
        break
```

```python
# Lesson 7: Strings in Python

language = "Python"

# Positive indexing (left to right, starts at 0)
print(language[0])      # P
print(language[2])      # t
print(language[5])      # n (last character, index = len-1)

# Negative indexing (right to left, starts at -1)
print(language[-1])     # n (last)
print(language[-3])     # h (third from end)

# Slicing
print(language[0:3])    # Pyt (stop 3 is excluded)
print(language[2:])     # thon (from index 2 to end)
print(language[:4])     # Pyth (from start to index 3)
print(language[::-1])   # nohtyP (reversed)
```
