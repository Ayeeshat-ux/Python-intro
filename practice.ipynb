```python
# CLASS ACTIVITY 3.3a
# Create one variable of each type for a product
# e.g., product name, price, quantity, availability

# int:
product_quantity = 40

# float:
product_price = 20.5

# str:
product_name = "Chair"

# bool:
product_available = True

# Print each with type():
print("Product Name:", product_name, "| Type:", type(product_name))
print("Product Price:", product_price, "| Type:", type(product_price))
print("Product Quantity:", product_quantity, "| Type:", type(product_quantity))
print("Product Availability:", product_available, "| Type:", type(product_available))
```

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

sum_result = num1 + num2
difference_result = num1 - num2
product_result = num1 * num2

print("\nResults:")
print("Sum:", sum_result)
print("Difference:", difference_result)
print("Product:", product_result)

print("\nData Types:")
print("Type of Sum:", type(sum_result))
print("Type of Difference:", type(difference_result))
print("Type of Product:", type(product_result))
```

```python
# practice question: Topic 3

# Collect the student information
name = input("Enter your name: ")
age = int(input("Enter your age: "))
gpa = float(input("Enter your GPA: "))

# Convert the user's answer into a boolean value
enrolled = input("Are you enrolled? (yes/no):")
enrolled_status = enrolled =="yes"


# Calculate the student's age on their next birthday
next_birthday_age = age + 1

# Round the GPA to one decimal place
rounded_gpa = round(gpa, 2)

print("\n" + "=" * 45)
print("   STUDENT PROFILE")
print("=" * 45)

print("Name:", name)
print("Type:", type(name))

print("Age:", age)
print("Type:", type(age))

print("GPA:", gpa)
print("Type:", type(gpa))

print("Enrolled:", enrolled_status)
print("Type:", type(enrolled_status))

print("-" * 45)

print("Age on next birthday:", next_birthday_age)
print("Rounded GPA:", rounded_gpa)

print("=" * 45)




```
