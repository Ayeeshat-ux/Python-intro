```python
# PRACTICE QUESTION — Topic 6
# Write your Number Pattern Printer here

# Step 1: Ask the user for a number between 1 and 20
number = int(input("Enter a number between 1 and 20: "))

# Validate the input using a while loop
while number < 1 or number > 20:
    print("Invalid input. Please enter a number between 1 and 20.")
    number = int(input("Enter a number between 1 and 20: "))

# Step 2: Print the descending number triangle
print("\nNumber Pattern:")

for row in range(number, 0, -1):
    for value in range(number, number - row, -1):
        print(value, end=" ")
    print()

# Step 3: Print only numbers that are NOT multiples of N
print("\nNumbers that are NOT multiples of", number)

count = 0
for i in range(1, (number * 2) + 1):

    if i % number == 0:
        continue

    print(i)

    count += 1

    if count == 5:
        break
```

```python
# Lesson 7: Strings in Python

language = "Python"

# Positive indexing (left to right, starts at 0)
print(language[0])      # P
print(language[2])      # t
print(language[5])      # n (last character, index = len-1)

# Negative indexing (right to left, starts at -1)
print(language[-1])     # n (last)
print(language[-3])     # h (third from end)

# Slicing
print(language[0:3])    # Pyt (stop 3 is excluded)
print(language[2:])     # thon (from index 2 to end)
print(language[:4])     # Pyth (from start to index 3)
print(language[::-1])   # nohtyP (reversed)
```

```python
#  Modifying lists

cart = ["bread", "milk", "eggs"]

# append — add to end
cart.append("butter")
print("After append:", cart)

# insert — add at position
cart.insert(1, "juice")       # insert at index 1
print("After insert:", cart)

# pop — remove at index (returns it)
removed = cart.pop(0)         # remove first item
print(f"Removed: {removed}, Cart: {cart}")

# remove — remove by value
cart.remove("milk")           # remove by value
print("After remove:", cart)
```

```python
# PRACTICE QUESTION — Topic 8
# Write your inventory system here

# Step 1: Create a list of 5 products (name, price, quantity)
products = [
    ("Rice", 40, 100),
    ("Beans", 20, 50),
    ("Sugar", 7, 30),
    ("Milk", 5, 25),
    ("Bread", 2, 30)
]

# Step 2: Add 2 more products
products.append(("Tea", 1, 50))
products.append(("Butter", 1, 60))

# Step 3: Remove one product by unpacking its name
for product in products:
    name, price, quantity = product
    if name == "Sugar":
        products.remove(product)
        break

# Step 4: Display the inventory using enumerate()
print("=" * 50)
print("          SHOP INVENTORY")
print("=" * 50)

for number, product in enumerate(products, start=1):
    name, price, quantity = product
    print(f"{number}. {name} - Price: ${price}, Quantity: {quantity}")

# Step 5: Calculate the total inventory value
total_value = 0

for product in products:
    name, price, quantity = product
    total_value += price * quantity

# Step 6: Find the most expensive product
most_expensive = products[0]

for product in products:
    if product[1] > most_expensive[1]:
        most_expensive = product

print("-" * 50)
print("Total Inventory Value:","$", total_value)
print("Most Expensive Item:", most_expensive[0], "-", "$", most_expensive[1])
print("=" * 50)
```

```python
# Lesson 9: Data Sets & Dictionaries

# Sets basics

# Duplicates are automatically removed
numbers = {1, 2, 3, 2, 1, 3, 4}
print(numbers)              # {1, 2, 3, 4} — duplicates gone

# Remove duplicates from a list
raw_list = ["python", "java", "python", "c++", "java"]
unique_languages = set(raw_list)
print(unique_languages)     # {'python', 'java', 'c++'}

# Membership testing (fast)
print("python" in unique_languages)    # True
print("ruby" in unique_languages)      # False

# Empty set
empty_set = set()           # NOT {}
print(type(empty_set))      # <class 'set'>
```

```python
 # Set methods

my_set = {1, 2, 3}
my_set.add(4)               # add a single item
my_set.add(2)               # already there — no change
print(my_set)               # {1, 2, 3, 4}

my_set.discard(3)           # remove if present (no error if absent)
my_set.remove(1)            # remove (raises KeyError if absent)
print(my_set)               # {2, 4}
```

```python
# Set operations

python_devs = {"Amara", "Kofi", "Ama", "Yaw"}
java_devs = {"Kofi", "Ama", "Abena", "Kojo"}

# Union — all developers
all_devs = python_devs | java_devs
print("All:", all_devs)

# Intersection — know both
both = python_devs & java_devs
print("Both Python and Java:", both)     # {'Kofi', 'Ama'}

# Difference — Python only
python_only = python_devs - java_devs
print("Python only:", python_only)       # {'Amara', 'Yaw'}

# Symmetric difference — not in both
exclusive = python_devs ^ java_devs
print("Not shared:", exclusive)
```

```python
# PRACTICE QUESTION — Topic 9
# Write your Course Enrollment Tracker here

# Step 1: Create a dictionary mapping courses to enrolled students
courses = {
    "Python": {"Aishat", "John", "Mary"},
    "Data Science": {"Aishat", "David", "Grace"},
    "SQL": {"John", "Sarah", "Michael"}
}

# Step 2: Add 2 more students to the Python course
courses["Python"].add("Bimbo")
courses["Python"].add("Esther")

# Step 3: Find students in both Python and Data Science
both_courses = courses["Python"].intersection(courses["Data Science"])

# Find students in Python only
python_only = courses["Python"].difference(courses["Data Science"])

# Step 4: Display each course and the number of students
print("=" * 50)
print("      COURSE ENROLLMENT TRACKER")
print("=" * 50)

for course, students in courses.items():
    print(f"{course}: {len(students)} students")

# Step 5: Find the course with the most students
largest_course = ""
largest_count = 0

for course, students in courses.items():
    if len(students) > largest_count:
        largest_count = len(students)
        largest_course = course

# Display the results
print("-" * 50)
print("Students in both Python and Data Science:", both_courses)
print("Students in Python only:", python_only)
print(f"Course with the most students: {largest_course} ({largest_count} students)")
print("=" * 50)
```

```python
# return stops the function

def check_pass(score):
    if score >= 50:
        return "PASS"    # function ends here if score >= 50
    return "FAIL"        # only reaches here if score < 50

print(check_pass(75))    # PASS
print(check_pass(40))    # FAIL

# Returning multiple related values as a tuple
def min_max(numbers):
    return min(numbers), max(numbers)

low, high = min_max([5, 2, 8, 1, 9])
print(f"Min: {low}, Max: {high}")
```

```python
# Define a function send_message(recipient, message, urgent=False)
# If urgent is True, print: "[URGENT] To: recipient — message"
# Otherwise print: "To: recipient — message"
# Call it 3 times: once with urgent=True, twice without specifying urgent.

# Your code here:

def send_message(recipient, message, urgent=False):
    if urgent:
        print("[URGENT] To:", recipient, "—", message)
    else:
        print("To:", recipient, "—", message)

# Call the function
send_message("Aishat", "Please submit your assignment.", urgent=True)
send_message("John", "The meeting starts at 10:00 AM.")
send_message("Mary", "Have a great day!")
```

```python
# Define a function calculate_discount(price, discount_percent=10)
# that returns the final price after the discount.
# Call it: once using the default 10%, once with 25%, once with 0%.

# Your code here:

def calculate_discount(price, discount_percent=10):
    discount_amount = price * (discount_percent / 100)
    final_price = price - discount_amount
    return final_price

# Call the function
print("Price after 10% discount:", calculate_discount(5000))
print("Price after 25% discount:", calculate_discount(5000, 25))
print("Price after 0% discount:", calculate_discount(5000, 0))
```

```python
# Local vs global scope

counter = 0          # global variable

def increment():
    global counter   # declare intent to modify the global
    counter += 1

increment()
increment()
print(counter)       # 2

def local_example():
    local_var = "I am local"
    print(local_var)     # works fine inside the function

local_example()
# print(local_var)   # NameError! local_var does not exist out here
```

```python
# Better practice — parameters instead of global

# BAD — relies on global state
total = 0
def add_to_total(amount):
    global total
    total += amount

add_to_total(50)
add_to_total(30)
print("Total (global):", total)    # 80

# BETTER — pass in, return out
def add_amount(current_total, amount):
    return current_total + amount

balance = 0
balance = add_amount(balance, 50)
balance = add_amount(balance, 30)
print("Balance (clean):", balance)  # 80
# No global state — easier to test and reuse
```

```python
# PRACTICE QUESTION — Topic 10
# Write your Student Grade Calculator here

# Function to collect three exam scores
def get_scores():
    score1 = float(input("Enter first exam score: "))
    score2 = float(input("Enter second exam score: "))
    score3 = float(input("Enter third exam score: "))
    return (score1, score2, score3)

# Function to calculate the average score
def calculate_average(scores):
    average = sum(scores) / len(scores)
    return average

# Function to classify the grade
def classify_grade(average, passing_score=50):
    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= passing_score:
        return "D"
    else:
        return "F"

# Function to display the student's report
def display_report(name, scores, average, grade):
    print("\n" + "=" * 45)
    print("        STUDENT GRADE REPORT")
    print("=" * 45)
    print("Name:", name)
    print("Scores:", scores)
    print("Average Score:", round(average, 2))
    print("Grade:", grade)
    print("=" * 45)

# Main Program
name = input("Enter your name: ")

scores = get_scores()
average = calculate_average(scores)
grade = classify_grade(average)

display_report(name, scores, average, grade)
```
