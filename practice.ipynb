```python
# PRACTICE QUESTION — Topic 6
# Write your Number Pattern Printer here

# Step 1: Ask the user for a number between 1 and 20
number = int(input("Enter a number between 1 and 20: "))

# Validate the input using a while loop
while number < 1 or number > 20:
    print("Invalid input. Please enter a number between 1 and 20.")
    number = int(input("Enter a number between 1 and 20: "))

# Step 2: Print the descending number triangle
print("\nNumber Pattern:")

for row in range(number, 0, -1):
    for value in range(number, number - row, -1):
        print(value, end=" ")
    print()

# Step 3: Print only numbers that are NOT multiples of N
print("\nNumbers that are NOT multiples of", number)

count = 0
for i in range(1, (number * 2) + 1):

    if i % number == 0:
        continue

    print(i)

    count += 1

    if count == 5:
        break
```

```python
# Lesson 7: Strings in Python

language = "Python"

# Positive indexing (left to right, starts at 0)
print(language[0])      # P
print(language[2])      # t
print(language[5])      # n (last character, index = len-1)

# Negative indexing (right to left, starts at -1)
print(language[-1])     # n (last)
print(language[-3])     # h (third from end)

# Slicing
print(language[0:3])    # Pyt (stop 3 is excluded)
print(language[2:])     # thon (from index 2 to end)
print(language[:4])     # Pyth (from start to index 3)
print(language[::-1])   # nohtyP (reversed)
```

```python
# Lesson 9: Data Sets & Dictionaries

# Sets basics

# Duplicates are automatically removed
numbers = {1, 2, 3, 2, 1, 3, 4}
print(numbers)              # {1, 2, 3, 4} — duplicates gone

# Remove duplicates from a list
raw_list = ["python", "java", "python", "c++", "java"]
unique_languages = set(raw_list)
print(unique_languages)     # {'python', 'java', 'c++'}

# Membership testing (fast)
print("python" in unique_languages)    # True
print("ruby" in unique_languages)      # False

# Empty set
empty_set = set()           # NOT {}
print(type(empty_set))      # <class 'set'>
```

```python
# Set operations

python_devs = {"Amara", "Kofi", "Ama", "Yaw"}
java_devs = {"Kofi", "Ama", "Abena", "Kojo"}

# Union — all developers
all_devs = python_devs | java_devs
print("All:", all_devs)

# Intersection — know both
both = python_devs & java_devs
print("Both Python and Java:", both)     # {'Kofi', 'Ama'}

# Difference — Python only
python_only = python_devs - java_devs
print("Python only:", python_only)       # {'Amara', 'Yaw'}

# Symmetric difference — not in both
exclusive = python_devs ^ java_devs
print("Not shared:", exclusive)
```

```python
highest_student = ""
highest_score = 0

lowest_student = ""
lowest_score = 100

with open("scores.txt", "r", encoding="utf-8") as f:
    for line in f:
        parts = line.strip().split()
        name = parts[0]
        score = int(parts[1])

        if score > highest_score:
            highest_score = score
            highest_student = name

        if score < lowest_score:
            lowest_score = score
            lowest_student = name

print("Highest score:", highest_student, "-", highest_score)
print("Lowest score:", lowest_student, "-", lowest_score)
```

```python
# with vs manual open/close

# WITHOUT with — risky
f = open('data.txt', 'w')
f.write('test')
# If an error happened here, f.close() would never run
f.close()

# WITH with — safe and clean
with open('data.txt', 'w', encoding='utf-8') as f:
    f.write('test')
# f is automatically closed here, guaranteed

# Verify the file is closed
print('File closed:', f.closed)   # True

# with for reading
with open('data.txt', 'r', encoding='utf-8') as f:
    content = f.read()
print(content)   # can use content here — file was auto-closed above
```

```python
# Multiple files in one with

# Create input file
with open('input.txt', 'w', encoding='utf-8') as f:
    f.write('apple\nbanana\nmango\n')

# Copy + transform: read one file, write to another — both open at once
with open('input.txt', 'r', encoding='utf-8') as src, \
     open('upper.txt', 'w', encoding='utf-8') as dst:
    for line in src:
        dst.write(line.upper())

with open('upper.txt', 'r') as f:
    print(f.read())
```

```python
# DictReader and DictWriter

import csv

students = [
    {'name': 'Amara', 'score': 85, 'grade': 'A'},
    {'name': 'Kofi',  'score': 72, 'grade': 'B'},
    {'name': 'Ama',   'score': 91, 'grade': 'A'},
]

# Write with DictWriter
with open('students.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=['name', 'score', 'grade'])
    writer.writeheader()
    writer.writerows(students)
    # Read with DictReader — access by column name
with open('students.csv', 'r', newline='', encoding='utf-8') as f:
    for row in csv.DictReader(f):
        print(f"{row['name']:10} scored {row['score']} — Grade {row['grade']}")
```

```python
import csv

# Create the CSV file and add 4 products
with open("products.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)

    # Header
    writer.writerow(["name", "price", "quantity"])

    # Products
    writer.writerow(["Laptop", 900, 2])
    writer.writerow(["Mouse", 25, 5])
    writer.writerow(["Keyboard", 50, 3])
    writer.writerow(["Monitor", 300, 2])


# Read the CSV file
with open("products.csv", "r", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)

    for product in reader:
        name = product["name"]
        price = float(product["price"])
        quantity = int(product["quantity"])

        total_value = price * quantity

        print(
            f"{name}: Price = ${price:.2f}, "
            f"Quantity = {quantity}, "
            f"Total Value = ${total_value:.2f}"
        )
```

```python
import csv

total_score = 0
count = 0
highest_score = -1
highest_student = ""
grade_a_count = 0

with open("students.csv", "r", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)

    for student in reader:
        name = student["name"]
        score = float(student["score"])
        grade = student["grade"]

        # Calculate total and count
        total_score += score
        count += 1

        # Find highest score
        if score > highest_score:
            highest_score = score
            highest_student = name

        # Count grade A
        if grade.upper() == "A":
            grade_a_count += 1

# Calculate average
average = total_score / count

print(f"Average score: {average:.2f}")
print(f"Highest score: {highest_student} - {highest_score:.2f}")
print(f"Number of students with grade A: {grade_a_count}")
```

```python
# PRACTICE QUESTION — Topic 13
# Simple Contact Book

import csv

# 1. Create contacts.csv with 3 initial contacts
with open("contacts.csv", "w", newline="", encoding="utf-8") as f:
    fieldnames = ["name", "phone", "email"]
    writer = csv.DictWriter(f, fieldnames=fieldnames)

    writer.writeheader()

    writer.writerow({
        "name": "Amara",
        "phone": "08012345678",
        "email": "amara@gmail.com"
    })

    writer.writerow({
        "name": "Kofi",
        "phone": "08023456789",
        "email": "kofi@gmail.com"
    })

    writer.writerow({
        "name": "Ama",
        "phone": "08034567890",
        "email": "ama@gmail.com"
    })


# 2. Function to add a contact
def add_contact(name, phone, email):
    with open("contacts.csv", "a", newline="", encoding="utf-8") as f:
        fieldnames = ["name", "phone", "email"]
        writer = csv.DictWriter(f, fieldnames=fieldnames)

        writer.writerow({
            "name": name,
            "phone": phone,
            "email": email
        })


# 3. Function to search for a contact
def search_contact(name):
    with open("contacts.csv", "r", newline="", encoding="utf-8") as f:
        reader = csv.DictReader(f)

        for contact in reader:
            if contact["name"].lower() == name.lower():
                return contact

    return None


# 4. Add 2 more contacts
add_contact("Yaw", "08045678901", "yaw@gmail.com")
add_contact("Adwoa", "08056789012", "adwoa@gmail.com")


# 5. Ask the user which contact to search for
search_name = input("Enter the name to search for: ")

result = search_contact(search_name)


# 6. Display the result
if result:
    print("\nContact found!")
    print("------------------------------")
    print("Name:", result["name"])
    print("Phone:", result["phone"])
    print("Email:", result["email"])
else:
    print("\nContact not found.")
```
