```python
# Lesson 4:  Python Operators & Expression

total_eggs = 250
eggs_per_cake = 3
price_per_cake = 12.50

# Calculate: complete cakes, leftover eggs, total revenue
# Calculate the number of complete cakes made
complete_cakes = total_eggs // eggs_per_cake

# Calculate the number of eggs left over
leftover_eggs = total_eggs % eggs_per_cake

# Calculate the total revenue
total_revenue = total_eggs * price_per_cake

print("complete cakes:", complete_cakes)
print("leftover eggs:", leftover_eggs)
print("total revenue: GHS", total_revenue)
```

```python
age = 22
has_id = True
is_banned = False

# 'and' — both must be true
can_enter = age >= 18 and has_id and not is_banned
print("Can enter venue:", can_enter)   # True

# 'or' — at least one must be true
needs_attention = age < 18 or is_banned
print("Needs attention:", needs_attention)  # False

# Correct way to check multiple values
grade = "B"
is_pass_grade = grade == "A" or grade == "B" or grade == "C"
print("Is pass grade:", is_pass_grade)  # True
```

```python
# Without parentheses — follows rules
result1 = 2 + 3 * 4      # 14  (* before +)
result2 = 2 ** 3 + 1     # 9   (** before +)
result3 = 10 - 2 * 3 + 1 # 5   (* first, then left to right)

print(result1, result2, result3)

# With parentheses — you control the order
result4 = (2 + 3) * 4    # 20
result5 = (2 ** 3) + 1   # 9  (same result, but clearer intent)
print(result4, result5)

# Complex expression — parentheses add clarity
score = 75
attendance = 0.9
# Is this student eligible? Score over 70 AND attendance over 80%
eligible = (score >= 70) and (attendance >= 0.8)
print("Eligible:", eligible)
```

```python
# PRACTICE QUESTION — Topic 4

# Step 1: Collect the student's scores
exam_score = float(input("Enter your exam score: "))
coursework_score = float(input("Enter your coursework score: "))
attendance = float(input("Enter your attendance percentage: "))

# Step 2: Calculate the final score
# Exam contributes 60% and coursework contributes 40%.
final_score = (exam_score * 0.60) + (coursework_score * 0.40)

# Step 3: Apply a 5% bonus if attendance is 90% or above
bonus_applied = False

if attendance >= 90:
    final_score += 5
    bonus_applied = True

# Step 4: Determine if the student passed
# The student must have a final score of at least 50 and attendance of at least 60%.
if final_score >= 50 and attendance >= 60:
    result = "Pass"
else:
    result = "Fail"

# Step 5: Display the results
print("\n" + "=" * 45)
print("         GRADE CALCULATOR")
print("=" * 45)
print("Final Score:", round(final_score, 2))
print("Pass/Fail Status:", result)
print("Attendance Bonus Applied:", bonus_applied)
print("=" * 45)
```

```python
weight = float(input("Enter weight (kg): "))
height = float(input("Enter height (m): "))

bmi = weight / (height ** 2)
print(f"Your BMI is: {bmi:.1f}")

#  Classify the BMI
if bmi < 18.5:
    print("Classification: Underweight")
elif bmi < 25:
    print("Classification: Normal")
elif bmi < 30:
    print("Classification: Overweight")
else:
    print("Classification: Obese")
```
