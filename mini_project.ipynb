```python
# MINI PROJECT 2: Smart Greeting System

# Step 1: Collect user information
# Get the user's name, age, city and time of day.
name = input("enter your name:")
age = int(input("enter your age:"))
city = input("enter your city:")
time = input("enter the time of day (Morning/Afternoon/Evening):")

# Step 2: Determine greeting based on time of day
# Display the correct greeting based on the user's input.
if time == "Morning":
    greeting = "Good Morning"
elif time == "Afternoon":
    greeting = "Good Afternoon"
else:
    greeting = "Good Evening"

# Step 3: Calculate birth year
# Subtract the user's age from 2024 to estimate the birth year.
birth_year = 2026 - age

# Step 4: Display formatted card
# Print the user's information in a neat card.
print("\n" + "=" * 45)
print("      SMART GREETING SYSTEM")
print("=" * 45)
print(greeting + ", " + name + "!")
print("age:", age)
print("city:", city)
print("year of birth:", birth_year)
print("-" * 45)
print("Thank you for using the Smart Greeting System!")
print("=" * 45)
```
