```python
# CUMULATIVE PROJECT — Topic 10
# Library Book Tracker

# Function to add a book to the library
def add_book(library, title, author, available=True):
    book = {
        "title": title,
        "author": author,
        "available": available
    }
    library.append(book)

# Function to search for a book by title
def search_book(library, title):
    for book in library:
        if book["title"] == title:
            return book
    return None

# Function to borrow a book
def borrow_book(library, title):
    book = search_book(library, title)

    if book is None:
        return "Book not found."

    if book["available"]:
        book["available"] = False
        return f"You borrowed '{title}'."
    else:
        return f"'{title}' is currently unavailable."

# Function to return a book
def return_book(library, title):
    book = search_book(library, title)

    if book is None:
        return "Book not found."

    if not book["available"]:
        book["available"] = True
        return f"You returned '{title}'."
    else:
        return f"'{title}' is already available."

# Function to display the library catalogue
def display_catalogue(library):
    print("\n" + "=" * 50)
    print("           LIBRARY CATALOGUE")
    print("=" * 50)

    for number, book in enumerate(library, start=1):
        status = "Available" if book["available"] else "Borrowed"

        print(f"{number}. Title: {book['title']}")
        print(f"   Author: {book['author']}")
        print(f"   Status: {status}")
        print("-" * 50)

# Main Program
library = []

# Add 4 books
add_book(library, "Python Basics", "John Smith")
add_book(library, "Data Science", "Jane Doe")
add_book(library, "SQL Essentials", "David Brown")
add_book(library, "Machine Learning", "Andrew Ng")

# Borrow 2 books
print(borrow_book(library, "Python Basics"))
print(borrow_book(library, "SQL Essentials"))

# Return 1 book
print(return_book(library, "Python Basics"))

# Display the full catalogue
display_catalogue(library)
```

```python
# CUMULATIVE PROJECT — Topic 13
# Student Records Manager

import csv

FILE_NAME = "student_records.csv"


# 1. Add a student
def add_student(name, score, grade):
    try:
        score = float(score)

        if score < 0 or score > 100:
            raise ValueError("Score must be between 0 and 100.")

        records = load_records()

        records.append({
            "name": name,
            "score": score,
            "grade": grade
        })

        save_records(records)

        print(f"Student '{name}' added successfully.")

    except ValueError as e:
        print(f"Error: {e}")


# 2. Load records from CSV
def load_records():
    records = []

    try:
        with open(FILE_NAME, "r", newline="", encoding="utf-8") as f:
            reader = csv.DictReader(f)

            for row in reader:
                records.append(row)

    except FileNotFoundError:
        # File doesn't exist yet, so start with an empty list
        return []

    except OSError as e:
        print(f"File error: {e}")

    return records


# 3. Save records to CSV
def save_records(records):
    try:
        with open(FILE_NAME, "w", newline="", encoding="utf-8") as f:
            fieldnames = ["name", "score", "grade"]

            writer = csv.DictWriter(
                f,
                fieldnames=fieldnames
            )

            writer.writeheader()
            writer.writerows(records)

    except OSError as e:
        print(f"File error while saving: {e}")


# 4. Search for a student
def search_student(name):
    records = load_records()

    for student in records:
        if student["name"].lower() == name.lower():
            return student

    return None


# 5. Display all students
def display_all(records):
    if not records:
        print("\nNo student records found.")
        return

    print("\n" + "=" * 50)
    print("              STUDENT RECORDS")
    print("=" * 50)

    total_score = 0

    for number, student in enumerate(records, start=1):
        score = float(student["score"])
        total_score += score

        print(
            f"{number}. "
            f"Name: {student['name']} | "
            f"Score: {score:.2f} | "
            f"Grade: {student['grade']}"
        )

    average = total_score / len(records)

    print("-" * 50)
    print(f"Class Average: {average:.2f}")
    print("=" * 50)


# 6. Determine grade
def get_grade(score):
    if score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"


# 7. Main program
records = load_records()

while True:
    print("\n" + "=" * 50)
    print("          STUDENT RECORDS MANAGER")
    print("=" * 50)
    print("1. Add Student")
    print("2. View All")
    print("3. Search")
    print("4. Show Average")
    print("5. Exit")
    print("=" * 50)

    choice = input("Enter your choice (1-5): ")

    # Add Student
    if choice == "1":
        name = input("Enter student name: ").strip()

        if not name:
            print("Error: Name cannot be empty.")
            continue

        try:
            score = float(input("Enter score (0-100): "))

            if score < 0 or score > 100:
                raise ValueError("Score must be between 0 and 100.")

            grade = get_grade(score)

            add_student(name, score, grade)

            # Reload records after adding
            records = load_records()

        except ValueError as e:
            print(f"Error: {e}")

    # View All
    elif choice == "2":
        records = load_records()
        display_all(records)

    # Search
    elif choice == "3":
        name = input("Enter student name to search: ").strip()

        student = search_student(name)

        if student:
            print("\nStudent Found")
            print("-" * 30)
            print("Name:", student["name"])
            print("Score:", student["score"])
            print("Grade:", student["grade"])
        else:
            print("Student not found.")

    # Show Average
    elif choice == "4":
        records = load_records()

        if records:
            total = sum(float(student["score"]) for student in records)
            average = total / len(records)

            print(f"\nClass Average: {average:.2f}")
        else:
            print("No records available.")

    # Exit
    elif choice == "5":
        records = load_records()
        save_records(records)

        print("\nRecords saved successfully.")
        print("Student Records Manager closed.")
        break

    # Invalid menu choice
    else:
        print("Invalid choice. Please enter a number from 1 to 5.")
```

```python
# CUMULATIVE PROJECT — Topic 15
# School Management System

import csv


# ==========================================
# 1. BASE CLASS: PERSON
# ==========================================

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Hello, my name is {self.name} and I am {self.age} years old.")

    def __str__(self):
        return f"{self.name} - Age: {self.age}"


# ==========================================
# 2. STUDENT CLASS
# ==========================================

class Student(Person):
    def __init__(self, name, age, student_id, grades=None):
        super().__init__(name, age)
        self.student_id = student_id
        self.__grades = grades if grades is not None else []

    def add_grade(self, score):
        if score < 0 or score > 100:
            raise ValueError("Grade must be between 0 and 100.")
        self.__grades.append(score)

    def get_average(self):
        if not self.__grades:
            return 0
        return sum(self.__grades) / len(self.__grades)

    @property
    def grades(self):
        return self.__grades

    def introduce(self):
        print(
            f"Hello, my name is {self.name}. "
            f"I am a student with ID {self.student_id}."
        )

    def __str__(self):
        return (
            f"Student: {self.name} | "
            f"ID: {self.student_id} | "
            f"Average: {self.get_average():.2f}"
        )


# ==========================================
# 3. TEACHER CLASS
# ==========================================

class Teacher(Person):
    def __init__(self, name, age, subject, salary):
        super().__init__(name, age)
        self.subject = subject
        self.__salary = salary

    @property
    def salary(self):
        return self.__salary

    def teach(self):
        print(f"{self.name} is teaching {self.subject}.")

    def introduce(self):
        print(
            f"Hello, my name is {self.name}. "
            f"I am a teacher of {self.subject}."
        )

    def __str__(self):
        return (
            f"Teacher: {self.name} | "
            f"Subject: {self.subject} | "
            f"Salary: {self.salary:.2f}"
        )


# ==========================================
# 4. CREATE 3 STUDENTS
# ==========================================

student1 = Student("Aishat", 23, "ST001")
student1.add_grade(85)
student1.add_grade(90)
student1.add_grade(80)

student2 = Student("Amara", 21, "ST002")
student2.add_grade(78)
student2.add_grade(82)
student2.add_grade(88)

student3 = Student("Kofi", 22, "ST003")
student3.add_grade(92)
student3.add_grade(95)
student3.add_grade(90)


# ==========================================
# CREATE 2 TEACHERS
# ==========================================

teacher1 = Teacher("Mr. Johnson", 45, "Physics", 250000)
teacher2 = Teacher("Mrs. Williams", 40, "Mathematics", 230000)


# ==========================================
# 5. STORE EVERYONE IN ONE LIST
# ==========================================

people = [
    student1,
    student2,
    student3,
    teacher1,
    teacher2
]


# ==========================================
# POLYMORPHISM
# ==========================================

print("=" * 55)
print("             SCHOOL MANAGEMENT SYSTEM")
print("=" * 55)

print("\nINTRODUCTIONS")
print("-" * 55)

for person in people:
    person.introduce()


# ==========================================
# DISPLAY STUDENTS
# ==========================================

print("\nSTUDENTS")
print("-" * 55)

students = [student1, student2, student3]

for student in students:
    print(student)
    print(f"Grades: {student.grades}")


# ==========================================
# DISPLAY TEACHERS
# ==========================================

print("\nTEACHERS")
print("-" * 55)

teachers = [teacher1, teacher2]

for teacher in teachers:
    print(teacher)


# ==========================================
# 6. FIND STUDENT WITH HIGHEST AVERAGE
# ==========================================

highest_student = max(students, key=lambda student: student.get_average())

print("\nHIGHEST PERFORMING STUDENT")
print("-" * 55)
print(f"Name: {highest_student.name}")
print(f"Student ID: {highest_student.student_id}")
print(f"Average: {highest_student.get_average():.2f}")


# ==========================================
# 7. SAVE ALL RECORDS TO CSV
# ==========================================

with open("school_records.csv", "w", newline="", encoding="utf-8") as file:

    writer = csv.writer(file)

    writer.writerow([
        "Type",
        "Name",
        "Age",
        "Student ID",
        "Subject",
        "Salary",
        "Grades",
        "Average"
    ])

    # Save students
    for student in students:
        writer.writerow([
            "Student",
            student.name,
            student.age,
            student.student_id,
            "",
            "",
            ", ".join(map(str, student.grades)),
            f"{student.get_average():.2f}"
        ])

    # Save teachers
    for teacher in teachers:
        writer.writerow([
            "Teacher",
            teacher.name,
            teacher.age,
            "",
            teacher.subject,
            teacher.salary,
            "",
            ""
        ])


print("\n" + "=" * 55)
print("Records saved successfully to school_records.csv")
print("=" * 55)
```
