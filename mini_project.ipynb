```python
# MINI PROJECT 2: Smart Greeting System

# Step 1: Collect user information
# Get the user's name, age, city and time of day.
name = input("enter your name:")
age = int(input("enter your age:"))
city = input("enter your city:")
time = input("enter the time of day (Morning/Afternoon/Evening):")

# Step 2: Determine greeting based on time of day
# Display the correct greeting based on the user's input.
if time == "Morning":
    greeting = "Good Morning"
elif time == "Afternoon":
    greeting = "Good Afternoon"
else:
    greeting = "Good Evening"

# Step 3: Calculate birth year
# Subtract the user's age from 2024 to estimate the birth year.
birth_year = 2026 - age

# Step 4: Display formatted card
# Print the user's information in a neat card.
print("\n" + "=" * 45)
print("      SMART GREETING SYSTEM")
print("=" * 45)
print(greeting + ", " + name + "!")
print("age:", age)
print("city:", city)
print("year of birth:", birth_year)
print("-" * 45)
print("Thank you for using the Smart Greeting System!")
print("=" * 45)
```

```python
# Mini Project: Product Inventory Entry System
# Write your solution here

# Collect product information from the user
product_name = input("enter product name:")
unit_price = float(input("enter unit price:"))
quantity_in_stock = int(input("enter quantity in stock:"))

# Convert the user's answer into a Boolean value
on_sale = input("Is the product on sale? (yes/no): ")
on_sale_status = on_sale == "yes"

# Calculate the total inventory value
total_inventory_value = unit_price * quantity_in_stock

# Display the product information and data types
print("\n" + "=" * 45)
print("      PRODUCT INVENTORY CARD")
print("=" * 45)

print("Product Name:", product_name)
print("Type:", type(product_name))

print("Unit Price:", unit_price)
print("Type:", type(unit_price))


print("Quantity in Stock:", quantity_in_stock)
print("Type:", type(quantity_in_stock))

print("On Sale:", on_sale_status)
print("Type:", type(on_sale_status))

print("-" * 45)
print("Total Inventory Value:", total_inventory_value)

# Display the discounted price if the product is on sale
if on_sale_status:
    discounted_price = unit_price * 0.9
    print("Discounted Price:", discounted_price)

print("=" * 45)
print("Thank you for using the Product Inventory Entry System!")
print("=" * 45)



```
