```python
# MINI PROJECT 2: Smart Greeting System

# Step 1: Collect user information
# Get the user's name, age, city and time of day.
name = input("enter your name:")
age = int(input("enter your age:"))
city = input("enter your city:")
time = input("enter the time of day (Morning/Afternoon/Evening):")

# Step 2: Determine greeting based on time of day
# Display the correct greeting based on the user's input.
if time == "Morning":
    greeting = "Good Morning"
elif time == "Afternoon":
    greeting = "Good Afternoon"
else:
    greeting = "Good Evening"

# Step 3: Calculate birth year
# Subtract the user's age from 2024 to estimate the birth year.
birth_year = 2026 - age

# Step 4: Display formatted card
# Print the user's information in a neat card.
print("\n" + "=" * 45)
print("      SMART GREETING SYSTEM")
print("=" * 45)
print(greeting + ", " + name + "!")
print("age:", age)
print("city:", city)
print("year of birth:", birth_year)
print("-" * 45)
print("Thank you for using the Smart Greeting System!")
print("=" * 45)
```

```python
# Mini Project: Product Inventory Entry System
# Write your solution here

# Collect product information from the user
product_name = input("enter product name:")
unit_price = float(input("enter unit price:"))
quantity_in_stock = int(input("enter quantity in stock:"))

# Convert the user's answer into a Boolean value
on_sale = input("Is the product on sale? (yes/no): ")
on_sale_status = on_sale == "yes"

# Calculate the total inventory value
total_inventory_value = unit_price * quantity_in_stock

# Display the product information and data types
print("\n" + "=" * 45)
print("      PRODUCT INVENTORY CARD")
print("=" * 45)

print("Product Name:", product_name)
print("Type:", type(product_name))

print("Unit Price:", unit_price)
print("Type:", type(unit_price))


print("Quantity in Stock:", quantity_in_stock)
print("Type:", type(quantity_in_stock))

print("On Sale:", on_sale_status)
print("Type:", type(on_sale_status))

print("-" * 45)
print("Total Inventory Value:", total_inventory_value)

# Display the discounted price if the product is on sale
if on_sale_status:
    discounted_price = unit_price * 0.9
    print("Discounted Price:", discounted_price)

print("=" * 45)
print("Thank you for using the Product Inventory Entry System!")
print("=" * 45)



```

```python
# CUMULATIVE PROJECT — Topic 10
# Library Book Tracker

# Function to add a book to the library
def add_book(library, title, author, available=True):
    book = {
        "title": title,
        "author": author,
        "available": available
    }
    library.append(book)

# Function to search for a book by title
def search_book(library, title):
    for book in library:
        if book["title"] == title:
            return book
    return None

# Function to borrow a book
def borrow_book(library, title):
    book = search_book(library, title)

    if book is None:
        return "Book not found."

    if book["available"]:
        book["available"] = False
        return f"You borrowed '{title}'."
    else:
        return f"'{title}' is currently unavailable."

# Function to return a book
def return_book(library, title):
    book = search_book(library, title)

    if book is None:
        return "Book not found."

    if not book["available"]:
        book["available"] = True
        return f"You returned '{title}'."
    else:
        return f"'{title}' is already available."

# Function to display the library catalogue
def display_catalogue(library):
    print("\n" + "=" * 50)
    print("           LIBRARY CATALOGUE")
    print("=" * 50)

    for number, book in enumerate(library, start=1):
        status = "Available" if book["available"] else "Borrowed"

        print(f"{number}. Title: {book['title']}")
        print(f"   Author: {book['author']}")
        print(f"   Status: {status}")
        print("-" * 50)

# Main Program
library = []

# Add 4 books
add_book(library, "Python Basics", "John Smith")
add_book(library, "Data Science", "Jane Doe")
add_book(library, "SQL Essentials", "David Brown")
add_book(library, "Machine Learning", "Andrew Ng")

# Borrow 2 books
print(borrow_book(library, "Python Basics"))
print(borrow_book(library, "SQL Essentials"))

# Return 1 book
print(return_book(library, "Python Basics"))

# Display the full catalogue
display_catalogue(library)
```
